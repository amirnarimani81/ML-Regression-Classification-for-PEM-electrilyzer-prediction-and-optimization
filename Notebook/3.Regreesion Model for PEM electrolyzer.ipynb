<hr style="border: 1px solid #008000;">

<p><b>1. Data Quality Issues:</b></p>
<ul>
  <li><b>Missing Values:</b> Fields like <i>cathode_area</i>, <i>Cell_voltage</i>, and <i>membrane_type</i> have missing or placeholder values (e.g., "nane").</li>
  <li><b>Inconsistent Formats:</b> Mixed use of formulas and raw values in <i>Hydrogen_flow_rate</i> and <i>Temprature_cell(k)</i>. Typos like "Temprature" and incorrect compound names like "ruthennium_oxide".</li>
  <li><b>Non-Standard Categories:</b> Duplicate entries like "iridium_oxide_(iro2)" and "platinum_and_ruthenium_(Ruo2)" complicate parsing.</li>
</ul>

<p><b>2. Structural Problems:</b></p>
<ul>
  <li><b>Redundant Fields:</b> <i>power(w)</i> is a derived column from <i>Cell_voltage × Cell_current</i>.</li>
  <li><b>High Cardinality:</b> Fields like <i>anode_type</i> and <i>cathode_type</i> have excessive unique values that may need grouping.</li>
  <li><b>Sparse Columns:</b> <i>Cell_design_number</i> lacks analytical value and <i>electrolysis_type</i> has little variance (mostly PEM).</li>
</ul>

<p><b>3. Feature Engineering Challenges:</b></p>
<ul>
  <li><b>Mixed Types:</b> Fields like <i>catholyte</i> combine chemical formulas with natural language text.</li>
  <li><b>Multicollinearity:</b> <i>Cell_voltage</i>, <i>Cell_current</i>, and <i>power(w)</i> are strongly interdependent.</li>
  <li><b>Encoding Needs:</b> Categorical variables like <i>membrane_type</i> must be encoded but have limited diversity.</li>
</ul>

<p><b>4. Target Variable Ambiguity:</b></p>
<ul>
  <li><b>No Defined Target:</b> The dataset lacks a clear output label. Candidates: <i>Hydrogen_flow_rate</i> (regression), or <i>power(w)</i> for energy optimization.</li>
  <li><b>Imbalance:</b> Overrepresentation of certain setups (e.g., titanium electrodes), underrepresentation of others.</li>
</ul>

<p><b>5. Experimental Noise:</b></p>
<ul>
  <li><b>Inconsistent Measurements:</b> Formula-based entries (e.g., =0.3×22.4×60000/1000000) add variability and reduce reproducibility.</li>
  <li><b>Outliers:</b> Extreme values in <i>power(w)</i> (e.g., 0.02 vs. 876) hint at scale differences or errors.</li>
</ul>

<hr style="border: 1px solid #008000;">

<p style="color:green; font-weight: bold;">Recommendations:</p>
<ul>
  <li><b>Data Cleaning:</b> Replace formulas with computed values, standardize labels, and drop redundant fields.</li>
  <li><b>Feature Engineering:</b> Group rare categories, encode categorical variables, and extract numeric values from strings.</li>
  <li><b>Define ML Objective:</b> Choose a target variable and filter incomplete rows.</li>
  <li><b>EDA:</b> Focus on correlation analysis and clustering of experimental setups.</li>
  <li><b>ML Strategy:</b> Apply Linear regression, Lassio, Ridge , elastic net regression; Logistic Regression  and KNN for binary classification of yield tiers.</li>

## Regression Evaluation Metrics


Here are three common evaluation metrics for regression problems:

**Mean Absolute Error** (MAE) is the mean of the absolute value of the errors:

$$\frac 1n\sum_{i=1}^n|y_i-\hat{y}_i|$$

**Mean Squared Error** (MSE) is the mean of the squared errors:

$$\frac 1n\sum_{i=1}^n(y_i-\hat{y}_i)^2$$

**Root Mean Squared Error** (RMSE) is the square root of the mean of the squared errors:

$$\sqrt{\frac 1n\sum_{i=1}^n(y_i-\hat{y}_i)^2}$$

Comparing these metrics:

- **MAE** is the easiest to understand, because it's the average error.
- **MSE** is more popular than MAE, because MSE "punishes" larger errors, which tends to be useful in the real world.
- **RMSE** is even more popular than MSE, because RMSE is interpretable in the "y" units.

All of these are **loss functions**, because we want to minimize them.

```python
# Create and train the model
lr_reg = LinearRegression()
lr_reg.fit(X_train, y_train)
pred_lr_reg = lr_reg.predict(X_test)

# Cross-Validation
cv_lr = cross_val_score(lr_reg, X, y, cv=5, scoring='neg_mean_squared_error')
cv_lr_mean = -np.mean(cv_lr)

# Residual Analysis
residual_lr = y_test - pred_lr_reg

# Performance Metrics
mse_lr = mean_squared_error(y_test, pred_lr_reg)
rmse_lr = np.sqrt(mse_lr)
mae_lr = mean_absolute_error(y_test, pred_lr_reg)
r2_lr = r2_score(y_test, pred_lr_reg)

# Print Results
print("Linear Regression Results:")
print("Mean Squared Error:", mse_lr)
print("Root Mean Squared Error:", rmse_lr)
print("Mean Absolute Error:", mae_lr)
print("R2 Score:", r2_lr*100)
print("Cross-Validation Mean MSE:", cv_lr_mean)


# Plotting Results
plt.figure(figsize=(6, 4))
plt.bar(['MAE', 'MSE', 'R²', 'RMSE'], [mae_lr, mse_lr, r2_lr , rmse_lr], color='brown')
plt.title('Linear Regression - Evaluation Metrics')
plt.show()
```

```python
# Function to evaluate the model
def evaluate_model(model, X_train, X_test, y_train, y_test):
    # Fit the model on the training data
    model.fit(X_train, y_train)
    
    # Make predictions on the test data
    y_pred = model.predict(X_test)
    
    # Calculate performance metrics
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    return mse, mae, r2, y_pred
```

```python
def train_val(y_train, y_train_pred, y_test, y_pred, model_name):
    
    scores = {model_name + "_train" : {"R2" : r2_score(y_train, y_train_pred),
            "mae" : mean_absolute_error(y_train, y_train_pred),
            "mse" : mean_squared_error(y_train, y_train_pred),
            "rmse": np.sqrt(mean_squared_error(y_train, y_train_pred))},
             
              model_name + "_test" : {"R2" : r2_score(y_test, y_pred),
            "mae" : mean_absolute_error(y_test, y_pred),
            "mse" : mean_squared_error(y_test, y_pred),
            "rmse": np.sqrt(mean_squared_error(y_test, y_pred))}}
    return pd.DataFrame(scores)
```

  <h1>Ridge Regression</h1>
  <p>
    Ridge Regression is a type of linear regression that includes <strong>L2 regularization</strong> to prevent overfitting and improve generalization.
  </p>

  <h2>1️⃣ Why Use Ridge Regression?</h2>
  <ul>
    <li><strong>Reduces Overfitting</strong> → Helps when you have multicollinearity (high correlation between features).</li>
    <li><strong>Works Well with Small Datasets</strong> → Adds regularization to avoid high variance.</li>
    <li><strong>Controls Complexity</strong> → The <code>alpha</code> parameter controls the strength of regularization.</li>
  </ul>

  <h2>Key Parameters</h2>
  <p>
    <strong><code>alpha</code></strong> → Controls regularization strength. A higher value forces coefficients to shrink more, reducing overfitting.
  </p>
  <p>
    <strong><code>solver</code></strong> → The algorithm used to optimize the regression.<br>
    Options: <code>'svd'</code>, <code>'cholesky'</code>, <code>'lsqr'</code>, <code>'sparse_cg'</code>, <code>'sag'</code>, <code>'saga'</code>
  </p>
  <p>
    <strong><code>fit_intercept</code></strong> → If <code>True</code>, the model includes an intercept (bias). If <code>False</code>, it forces the regression through the origin.
  </p>
  <p>
    <strong><code>max_iter</code></strong> → Maximum number of iterations for convergence (only relevant for iterative solvers like <code>'saga'</code> and <code>'sag'</code>).
  </p>
  <p>
    <strong><code>tol</code></strong> → Tolerance for stopping criteria; a lower value results in more precise optimization but takes longer.
  </p>

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.metrics import make_scorer, mean_squared_error
```

```python
# 1) define models
# 1) define models
models = {
    "Linear Regression": LinearRegression(),
    "Polynomial Regression (deg 2)": Pipeline([
        ("poly", PolynomialFeatures(degree=2)),
        ("lr", LinearRegression())
    ]),
    "Lasso": Lasso(alpha=0.1, random_state=42),
    "Ridge": Ridge(alpha=1.0, random_state=42),
    "ElasticNet": ElasticNet(alpha=0.1, l1_ratio=0.5, random_state=42)
}

# 2) prepare storage
ks = range(2, 11)
results = {name: {'mse': [], 'rmse': [], 'r2': []} for name in models}

# 3) loop over k
for k in ks:
    cv = KFold(n_splits=k, shuffle=True, random_state=42)
    for name, model in models.items():
        cvres = cross_validate(
            model, X_train, y_train, cv=cv,
            scoring={
                'r2': 'r2',
                'mse': make_scorer(mean_squared_error)
            },
            return_train_score=False
        )
        mse_mean = np.mean(cvres['test_mse'])
        results[name]['mse'].append(mse_mean)
        results[name]['rmse'].append(np.sqrt(mse_mean))
        results[name]['r2'].append(np.mean(cvres['test_r2']))

# 4) plot each metric
for metric in ['mse', 'rmse', 'r2']:
    plt.figure(figsize=(8, 4))
    for name in models:
        plt.plot(ks, results[name][metric], marker='o', label=name)
    plt.title(f"{metric.upper()} vs. Number of Folds")
    plt.xlabel("k (folds)")
    plt.ylabel(metric.upper())
    plt.xticks(ks)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
records = []
for name, model in models.items():
    for k in range(2, 11):                     # k = 2…10
        cv = KFold(n_splits=k, shuffle=True, random_state=42)
        scoring = {'mse': 'neg_mean_squared_error', 'r2': 'r2'}
        scores = cross_validate(
            model, X_train, y_train,
            cv=cv, scoring=scoring, return_train_score=False
        )
        mse   = -scores['test_mse'].mean()     # negate because sklearn reports negative MSE
        rmse  = np.sqrt(mse)
        r2    =  scores['test_r2'].mean()
        records.append({
            'Model':   name,
            'K-Folds': k,
            'MSE':     mse,
            'RMSE':    rmse,
            'R2':      r2
        })

# — 4) Build and show the table —
cv_table = pd.DataFrame.from_records(records)
print(cv_table)
```
